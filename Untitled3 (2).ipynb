# Coffee Machine Project using:
# Abstract Class
# Encapsulation
# Inheritance
# Polymorphism

from abc import ABC, abstractmethod


# ---------------- ABSTRACT CLASS ----------------
class Coffee(ABC):

    @abstractmethod
    def recipe(self):
        pass

    @abstractmethod
    def show_image(self):
        pass


# ---------------- INHERITANCE + ENCAPSULATION ----------------
class Latte(Coffee):

    def __init__(self):
        self.__milk = "200ml Milk"
        self.__coffee = "1 Shot Espresso"
        self.__foam = "Milk Foam"

    def recipe(self):
        print("\nLatte Recipe:")
        print(self.__milk)
        print(self.__coffee)
        print(self.__foam)

    def show_image(self):
        print(r"""
          ( (
           ) )
        ........
        | LATTE |
        |      |
        \______/
         \____/
        """)


class Matcha(Coffee):

    def __init__(self):
        self.__matcha = "2 tsp Matcha Powder"
        self.__milk = "Hot Milk"
        self.__sugar = "Sugar"

    def recipe(self):
        print("\nMatcha Recipe:")
        print(self.__matcha)
        print(self.__milk)
        print(self.__sugar)

    def show_image(self):
        print(r"""
          ( (
           ) )
        .........
        |MATCHA |
        |       |
        \_______/
         \_____/
        """)


class Americano(Coffee):

    def __init__(self):
        self.__water = "Hot Water"
        self.__espresso = "2 Espresso Shots"

    def recipe(self):
        print("\nAmericano Recipe:")
        print(self.__water)
        print(self.__espresso)

    def show_image(self):
        print(r"""
          ( (
           ) )
        ............
        |AMERICANO |
        |          |
        \__________/
         \________/
        """)


# ---------------- COFFEE MACHINE ----------------
class CoffeeMachine:

    def make_coffee(self, choice):

        if choice == 1:
            coffee = Latte()

        elif choice == 2:
            coffee = Matcha()

        elif choice == 3:
            coffee = Americano()

        else:
            print("Invalid Choice")
            return

        coffee.show_image()
        coffee.recipe()


# ---------------- MAIN PROGRAM ----------------
machine = CoffeeMachine()

while True:

    print("\n===== COFFEE MACHINE =====")
    print("1. Latte")
    print("2. Matcha")
    print("3. Americano")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 4:
        print("Thank You for using Coffee Machine!")
        break

    machine.make_coffee(choice)
